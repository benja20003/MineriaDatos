## Actividad: Clases geométricas en Python

En esta actividad, vamos a definir dos clases en Python, "Cuadrado" y "Rectangulo", la clase "Cuadrado" heredará de la clase "Rectangulo". La clase "Rectangulo" tendrá dos atributos, "ancho" y "alto", que representan las dimensiones del rectángulo. La clase "Cuadrado" tendrá un atributo "lado" que representa la longitud del lado del cuadrado. En la clase "Cuadrado", se sobrescribirá el método ```__init__``` para inicializar el atributo "lado", y se asegurará que "ancho" y "alto" sean iguales. Y se implementará el método ```__str__``` para imprimir los atributos del cuadrado.
