### Actividad: Polimorfismo y encapsulamiento en Python

En esta actividad, vamos a definir una clase "Persona", vamos a implementar polimorfismo en la clase "Persona" con los métodos ```saludar``` y ```despedir```. Luego, vamos a implementar encapsulamiento en la clase "Persona" con los atributos "nombre" y "edad". Los atributos "nombre" y "edad" serán privados y solo se podrán acceder a ellos a través de métodos de acceso (getters y setters).
