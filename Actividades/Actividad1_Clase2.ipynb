### Actividad: Definir la clase "Rectangulo" en Python

En esta actividad, vamos a definir una clase llamada "Rectangulo" en Python. La clase "Rectangulo" tendrá dos atributos, "ancho" y "alto", que representan las dimensiones del rectángulo.

- Crear al menos dos objetos de la clase "Rectangulo"
- Calcular el área y el perímetro de los rectángulos
- Acceder y modificar los atributos de los rectángulos
- Imprimir los resultados

```python
#Creamos la clase "Rectangulo"

class Rectangulo:
    def __init__ (self, ancho, alto):
        self.ancho = ancho
        self.alto = alto
        
        
#Creamos los dos objetos

rectangulo1 = Rectangulo(10, 5)
rectangulo2 = Rectangulo(3, 7)

#Calculamos el area y perimetro de cada rectangulo
area1 = rectangulo1.ancho * rectangulo1.alto
perimetro1 = 2 * (rectangulo1.ancho + rectangulo1.alto)

area2 = rectangulo2.ancho * rectangulo2.alto
perimetro2 = 2 * (rectangulo2.ancho + rectangulo2.alto)

#Mostramos los resultados
print("Area del rectangulo 1:", area1)
print("Perimetro del rectangulo 1:", perimetro1)
print("Area del rectangulo 2:", area2)
print("Perimetro del rectangulo 2:", perimetro2)

print("Nuevos valores")

#Modificamos los atributos del rectangulo1 y rectangulo2
rectangulo1.ancho = 15
rectangulo1.alto = 10
rectangulo2.ancho = 5
rectangulo2.alto = 8

#Calculamos el area y perimetro de cada rectangulo con los nuevos valores
area1 = rectangulo1.ancho * rectangulo1.alto    
perimetro1 = 2 * (rectangulo1.ancho + rectangulo1.alto)

area2 = rectangulo2.ancho * rectangulo2.alto
perimetro2 = 2 * (rectangulo2.ancho + rectangulo2.alto)

#Mostramos los resultados con los nuevos valores
print("Area del rectangulo 1 con nuevos valores:", area1)
print("Perimetro del rectangulo 1 con nuevos valores:", perimetro1)
print("Area del rectangulo 2 con nuevos valores:", area2)
print("Perimetro del rectangulo 2 con nuevos valores:", perimetro2)


```
