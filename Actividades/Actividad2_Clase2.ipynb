## Actividad: Definir la clase "Rectangulo" con métodos en Python

En esta actividad, vamos a definir la clase "Rectangulo" en Python con métodos para calcular el área y el perímetro del rectángulo. La clase "Rectangulo" tendrá dos atributos, "ancho" y "alto", que representan las dimensiones del rectángulo.

- A la clase "Rectangulo" se le agregarán los métodos ```calcular_area``` y ```calcular_perimetro``` para calcular el área y el perímetro del rectángulo, respectivamente.
- Implementar el método especial ```__str__``` para imprimir los atributos del rectángulo.


````python
#Creamos la clase rectangulo con metodos y __str__

class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto
        
    #metodo para calcular el area del rectangulo
    def calcular_area(self):
        return self.ancho * self.alto

    #metodo para calcular el perimetro del rectangulo
    def calcular_perimetro(self):
        return 2 * (self.ancho + self.alto)

    #metodo para representar el objeto en texto
    def __str__(self):
        return f"Rectangulo de ancho {self.ancho} y alto {self.alto}"
    

#Creamos los objetos de prueba
rectangulo1 = Rectangulo(4, 2)
rectangulo2 = Rectangulo(5, 3)

#Imprimimos usando el metodo __str__ y los metodos de area y perimetro
print(rectangulo1)
print(f"Area: {rectangulo1.calcular_area()}")
print(f"Perimetro: {rectangulo1.calcular_perimetro()}")
print()
print(rectangulo2)
print(f"Area: {rectangulo2.calcular_area()}")
print(f"Perimetro: {rectangulo2.calcular_perimetro()}")


#Probamos si modificando el atributo el metodo se cambia solo

rectangulo1.ancho = 6
rectangulo1.alto = 4
rectangulo2.ancho = 7
rectangulo2.alto = 5

print()
print("Nuevos valores")
print(rectangulo1)
print(f"Area: {rectangulo1.calcular_area()}")
print(f"Perimetro: {rectangulo1.calcular_perimetro()}")
print()
print(rectangulo2)
print(f"Area: {rectangulo2.calcular_area()}")
print(f"Perimetro: {rectangulo2.calcular_perimetro()}")


````
