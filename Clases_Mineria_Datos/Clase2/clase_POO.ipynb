# Progamación Orientada a Objetos

## ¿Qué es la programación orientada a objetos?

La programación orientada a objetos (POO) es un paradigma de programación que utiliza "objetos" para diseñar aplicaciones y programas informáticos. Es un enfoque de modelado de software que organiza los datos en objetos y describe cómo los objetos pueden interactuar entre sí para resolver problemas de programación.

### Paradigma de programación orientada a objetos

Un paradigma de programación es un estilo de programación o un enfoque para la construcción de aplicaciones. La programación orientada a objetos es un paradigma de programación que se basa en el concepto de "objetos", que pueden contener datos en forma de campos, a menudo conocidos como atributos; y código, en forma de procedimientos, a menudo conocidos como métodos.

### Modela entidades del mundo real como objetos

La programación orientada a objetos modela entidades del mundo real como objetos. Por ejemplo, un objeto "coche" puede tener atributos como "marca", "modelo" y "color"; y métodos como "acelerar", "frenar" y "girar". Los objetos pueden interactuar entre sí a través de mensajes, lo que permite que los objetos se comuniquen y colaboren para resolver problemas de programación.

### Los objetos tienen atributos (propiedades) y métodos (comportamiento)

Los objetos tienen atributos, que son propiedades que describen el objeto, y métodos, que son comportamientos que el objeto puede realizar. Por ejemplo, un objeto "coche" puede tener atributos como "marca", "modelo" y "color"; y métodos como "acelerar", "frenar" y "girar". Los atributos y métodos de un objeto están definidos por su clase, que es una plantilla para objetos de cierto tipo.

## Beneficios de la programación orientada a objetos

La programación orientada a objetos tiene varios beneficios, incluyendo:

- **Modularidad**: Los objetos pueden ser reutilizados en diferentes partes de una aplicación o programa.
- **Encapsulación**: Los objetos ocultan su implementación interna y solo exponen una interfaz para interactuar con ellos.
- **Herencia**: Los objetos pueden heredar atributos y métodos de otras clases, lo que permite la reutilización de código.
- **Polimorfismo**: Los objetos pueden ser tratados como instancias de su clase base, lo que permite la flexibilidad en el diseño de aplicaciones.
- **Abstracción**: Los objetos representan entidades del mundo real y abstraen los detalles de implementación.

## Conceptos clave de la programación orientada a objetos

Algunos conceptos clave de la programación orientada a objetos incluyen:

- **Clase**: Una clase es una plantilla para objetos. Define atributos y métodos comunes a todos los objetos de cierto tipo.

- **Objeto**: Un objeto es una instancia de una clas. Cuando se crea un objeto, se asigna memoria para el objeto y se invoca el constructor de la clase.

- **Atributo**: Un atributo es una propiedad de un objeto. Los atributos son variables que contienen datos. Los atributos definen las propiedades de un objeto.

- **Método**: Un método es una función que pertenece a una clase. Los métodos definen el comportamiento de un objeto.

- **Herencia**: La herencia es un mecanismo que permite que una clase herede atributos y métodos de otra clase.

- **Polimorfismo**: El polimorfismo es un mecanismo que permite que un objeto se comporte de diferentes maneras en función del contexto.

- **Encapsulamiento**:El encapsulamiento es un mecanismo que oculta la implementación de un objeto y solo expone una interfaz.

- **Constructor**: Un constructor es un método especial que se llama cuando se crea un objeto.

## Ejemplo de programación orientada a objetos

````python
# Ejemplo: Modelar entidades del mundo real con clases

class Persona:
    def __init__(self, nombre1, edad):
        self.nombre = nombre1
        self.edad = edad

    def saludar(self):
        print(f'Hola, soy {self.nombre} y tengo {self.edad} años')
        
    def __str__(self):
        return f'Persona({self.nombre}, {self.edad})'
    
    def __repr__(self):
        return f'Persona({self.nombre}, {self.edad})'
    
class Estudiante(Persona):
    
    def __init__(self, nombre, edad, carrera):
        super().__init__(nombre, edad)
        self.carrera = carrera
        
    def saludar(self):
        print(f'Hola, soy {self.nombre} y estudio {self.carrera}')
        
    def __str__(self):
        return f'Estudiante({self.nombre}, {self.edad}, {self.carrera})'
    
    def __repr__(self):
        return f'Estudiante({self.nombre}, {self.edad}, {self.carrera})'
    
class Profesor(Persona):
        
        def __init__(self, nombre, edad, especialidad):
            super().__init__(nombre, edad)
            self.especialidad = especialidad
            
        def saludar(self):
            print(f'Hola, soy {self.nombre} y enseño {self.especialidad}')
            
        def __str__(self):
            return f'Profesor({self.nombre}, {self.edad}, {self.especialidad})'
        
        def __repr__(self):
            return f'Profesor({self.nombre}, {self.edad}, {self.especialidad})'
        
p1 = Persona('Juan', 30)

p2 = Estudiante('Ana', 20, 'Ingeniería')

p3 = Profesor('Luis', 40, 'Matemáticas')
````

````python
# Otro ejemplo

class Vehiculo:
    def __init__(self, marca, modelo, anio):
        self.marca = marca
        self.modelo = modelo
        self.anio = anio
        
    def __str__(self):
        return f'Vehiculo({self.marca}, {self.modelo}, {self.anio})'
    
    def __repr__(self):
        return f'Vehiculo({self.marca}, {self.modelo}, {self.anio})'
    
class Auto(Vehiculo):
    
    def __init__(self, marca, modelo, anio, color):
        super().__init__(marca, modelo, anio)
        self.color = color
        
    def __str__(self):
        return f'Auto({self.marca}, {self.modelo}, {self.anio}, {self.color})'
    
    def __repr__(self):
        return f'Auto({self.marca}, {self.modelo}, {self.anio}, {self.color})'
    
class Moto(Vehiculo):
    
    def __init__(self, marca, modelo, anio, cilindrada):
        super().__init__(marca, modelo, anio)
        self.cilindrada = cilindrada
        
    def __str__(self):
        return f'Moto({self.marca}, {self.modelo}, {self.anio}, {self.cilindrada})'
    
    def __repr__(self):
        return f'Moto({self.marca}, {self.modelo}, {self.anio}, {self.cilindrada})'
    
v1 = Vehiculo('Toyota', 'Corolla', 2019)

v2 = Auto('Ford', 'Fiesta', 2018, 'Rojo')

v3 = Moto('Honda', 'CBR', 2020, 250)
````

````python
# Ejemplo de programación orientada a objetos

class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
        
    def area(self):
        return self.base * self.altura
    
    def perimetro(self):
        return 2 * (self.base + self.altura)
    
r1 = Rectangulo(10, 5)

print(r1.area())
````

## Clases y Objetos en Python

En Python, las clases se definen utilizando la palabra clave `class`, seguida del nombre de la clase y dos puntos. Los métodos de una clase se definen como funciones dentro de la clase. ### Definición de una clase en Python

````python
class Car:
    def __init__(self, brand, model, color):
        self.brand = brand
        self.model = model
        self.color = color

    def accelerate(self):
        print(f"{self.brand} {self.model} is accelerating")

    def brake(self):
        print(f"{self.brand} {self.model} is braking")

    def turn(self, direction):
        print(f"{self.brand} {self.model} is turning {direction}")
````

## Métodos en Python

En Python, los métodos son funciones que pertenecen a una clase. Los métodos se definen dentro de una clase y se utilizan para definir el comportamiento de los objetos de la clase. Los métodos pueden acceder a los atributos de un objeto utilizando la notación de punto.

### Definición de métodos en Python

- Los métodos se definen como funciones dentro de una clase
- Se utilizan para definir el comportamiento de los objetos de la clase
- Sintaxis: ```def nombre_metodo(self, param1, param2, ...):```

### El parámetro "self"

- El parámetro "self" se utiliza para hacer referencia al objeto actual

### Métodos de instancia y métodos de clase

## Herencia en Python

La herencia es un mecanismo que permite que una clase herede atributos y métodos de otra clase. La clase que hereda se conoce como clase derivada o subclase, y la clase de la que hereda se conoce como clase base o superclase. La herencia permite la reutilización de código y la creación de jerarquías de clases.

### Definición de Herencia en Python

- La herencia se define colocando el nombre de la clase base entre paréntesis después del nombre de la clase derivada.
- Permite que la clase derivada herede atributos y métodos de la clase base.
- Reutiliza el código de la clase base en la clase derivada.

### Clase Base y Clase Derivada

- La clase base es la clase de la que hereda la clase derivada.
- La clase derivada es la clase que hereda de la clase base.
- Clase base (superclase): la clase de la que se hereda.
- Clase derivada (subclase): la clase que hereda de la clase base.

### Sintaxis para definir herencia

- ```class NombreClaseDerivada(NombreClaseBase):```
- La clase derivada hereda de la clase base. Esto significa que la clase derivada tiene todos los atributos y métodos de la clase base.

### Sobreescritura de métodos

- La sobreescritura de métodos es un mecanismo que permite que una clase derivada reemplace un método de la clase base.
- Una subclase puede sobrescribir (o reemplazar) un método de la superclase.
- La subclase puede proporcionar una implementación diferente del método.
- Se utiliza para extender o modificar el comportamiento heredado.

### Uso de la función ```super()```

- La función ```super()``` se utiliza para llamar a un método de la clase base desde la clase derivada.
- Sintaxis: ```super().nombre_metodo(argumentos)```

## Polimorfismo y encapsulamiento en Python

El polimorfismo es un mecanismo que permite que un objeto se comporte de diferentes maneras en función del contexto. En Python, el polimorfismo se logra mediante la definición de métodos con el mismo nombre en diferentes clases. El polimorfismo permite que un objeto se comporte de manera diferente en función de su tipo o clase.

### Definición de Polimorfismo en Python

- Objetos de diferentes clases pueden responder al mismo método o propiedad de manera diferente.
- Permite que un objeto se comporte de diferentes maneras en función de su tipo o clase.
- Trata a los objetos de diferentes clases de manera uniforme.
- Se logra mediante la definición de métodos con el mismo nombre en diferentes clases.
- Permite la reutilización de código y la creación de jerarquías de clases.
- Ejemplo: ```def area(self):```

### Ventajas del Polimorfismo

- Código más genérico y reutilizable.
- Mayor flexibilidad en el diseño de aplicaciones.
- Facilita la creación de jerarquías de clases.
- Facilita el mantenimiento y la extensión del código.

### Ejemplo de Polimorfismo en Python

````python
# Ejemplo de polimorfismo

class Animal:
    
    def __init__(self, nombre):
        self.nombre = nombre
        
    def hablar(self):
        pass
    
class Perro(Animal):
    
    def hablar(self):
        return 'Guau!'
    
class Gato(Animal):
    
    def hablar(self):
        return 'Miau!'
    
class Vaca(Animal):
    
    def hablar(self):
        return 'Muu!'
    
p = Perro('Rex')
g = Gato('Misi')
v = Vaca('Lola')

print(p.hablar())
print(g.hablar())
print(v.hablar())
````

### Encapsulamiento en Python

- El encapsulamiento es un mecanismo que oculta la implementación de un objeto y solo expone una interfaz.
- Permite que los atributos y métodos de un objeto sean privados o protegidos.
- Los atributos y métodos privados no se pueden acceder directamente desde fuera de la clase.
- Controla el acceso a los atributos y métodos de un objeto desde fuera de la clase.

### Niveles de Encapsulamiento en Python

- Público: Los atributos y métodos públicos se pueden acceder desde fuera de la clase.
- Protegido: Los atributos y métodos protegidos solo se pueden acceder desde la clase y sus subclases.
- Privado: Los atributos y métodos privados solo se pueden acceder desde la clase.

### Uso de propiedades (getters y setters)

- Las propiedades se utilizan para controlar el acceso a los atributos de un objeto.
- Se utilizan para definir métodos de acceso (getters) y métodos de modificación (setters).
- Permiten validar y modificar los valores de los atributos de un objeto.
- Sintaxis: ```@property``` y ```@atributo.setter```
- Ejemplo de uso en clase rectángulo

````python
# Ejemplo de uso de getters y setters en clase Rectangulo

class Rectangulo:
    def __init__(self, ancho, alto):
        self._ancho = ancho  # Atributo protegido
        self._alto = alto  # Atributo protegido

    @property
    def ancho(self):
        """Getter para obtener el valor de ancho"""
        return self._ancho

    @ancho.setter
    def ancho(self, nuevo_ancho):
        """Setter para modificar el valor de ancho"""
        if nuevo_ancho > 0:
            self._ancho = nuevo_ancho
        else:
            print("El ancho debe ser un valor positivo")

    @property
    def alto(self):
        """Getter para obtener el valor de alto"""
        return self._alto

    @alto.setter
    def alto(self, nuevo_alto):
        """Setter para modificar el valor de alto"""
        if nuevo_alto > 0:
            self._alto = nuevo_alto
        else:
            print("El alto debe ser un valor positivo")

    def calcular_area(self):
        """Método para calcular el área del rectángulo"""
        return self._ancho * self._alto

# Uso de la clase Rectangulo
rectangulo = Rectangulo(5, 10)
print(f"Área inicial: {rectangulo.calcular_area()}")  # Salida: Área inicial: 50

# Acceder a los atributos usando las propiedades
print(f"Ancho: {rectangulo.ancho}")  # Salida: Ancho: 5
print(f"Alto: {rectangulo.alto}")  # Salida: Alto: 10

# Modificar los atributos usando las propiedades
rectangulo.ancho = 8
rectangulo.alto = 6
print(f"Nueva área: {rectangulo.calcular_area()}")  # Salida: Nueva área: 48
````

## Estructura de un script enfocado a objetos en Python
