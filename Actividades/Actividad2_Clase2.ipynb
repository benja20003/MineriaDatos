## Actividad: Definir la clase "Rectangulo" con métodos en Python

En esta actividad, vamos a definir la clase "Rectangulo" en Python con métodos para calcular el área y el perímetro del rectángulo. La clase "Rectangulo" tendrá dos atributos, "ancho" y "alto", que representan las dimensiones del rectángulo.

- A la clase "Rectangulo" se le agregarán los métodos ```calcular_area``` y ```calcular_perimetro``` para calcular el área y el perímetro del rectángulo, respectivamente.
- Implementar el método especial ```__str__``` para imprimir los atributos del rectángulo.
